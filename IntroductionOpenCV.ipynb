```python
# Importing the OpenCV library
import cv2
# Reading the image using imread() function
image = cv2.imread('bus.jpg')

# Extracting the height and width of an image
h, w = image.shape[:2]
# Displaying the height and width
print("Height = {}, Width = {}".format(h, w))

cv2.imshow("Image", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import numpy as np
imgArr = np.asarray(image)
print(imgArr.shape)
```

```python
# Extracting RGB values.
# Here we have randomly chosen a pixel
# by passing in 100, 100 for height and width.
(B, G, R) = image[100, 100]

# Displaying the pixel values
print("R = {}, G = {}, B = {}".format(R, G, B))

# We can also pass the channel to extract
# the value for a specific channel
B = image[100, 100, 0]
print("B = {}".format(B))
```

```python
# We will calculate the region of interest
# by slicing the pixels of the image
roi = image[100 : 500, 200 : 700]
print(roi.shape)
cv2.imshow("ROI", roi)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# resize() function takes 2 parameters,
# the image and the dimensions
resize = cv2.resize(image, (500, 500))
print(resize.shape)
cv2.imshow("Resized Image", resize)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Define contrast and brightness
alpha = 1.5  # Simple contrast control (>1 increases contrast, <1 decreases)
beta = 0     # Simple brightness control (positive values increase brightness)

# Apply the transformation
image_resize = cv2.resize(image, (int(w/3), int(h/3)))
image_contrast = cv2.resize(cv2.convertScaleAbs(image, alpha=alpha, beta=beta), (int(w/3), int(h/3))) 
image_blur = cv2.resize(cv2.GaussianBlur(image, (7, 7), 0), (int(w/3), int(h/3)))
concatenated_image = cv2.hconcat([image_resize, image_blur, image_contrast])
cv2.imshow('Images (original, blur, contrast)', concatenated_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Get the dimensions of the image
(h, w) = image.shape[:2]

# Find the center of the image
center = (w // 2, h // 2)

# Define the rotation angle (e.g., 45 degrees)
angle = 45

# Define the scaling factor (1.0 means no scaling)
scale = 1.0

# Generate the rotation matrix
rotation_matrix = cv2.getRotationMatrix2D(center, angle, scale)

# Apply the rotation using cv2.warpAffine
rotated_image = cv2.warpAffine(image, rotation_matrix, (w, h))

# Show the rotated image (or save it)
cv2.imshow('Rotated Image', rotated_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import numpy as np

# Calculate the new bounding dimensions after rotation
# Convert the angle from degrees to radians
angle_rad = np.radians(angle)
new_w = int(abs(w * np.cos(angle_rad)) + abs(h * np.sin(angle_rad)))
new_h = int(abs(w * np.sin(angle_rad)) + abs(h * np.cos(angle_rad)))

# Generate the rotation matrix with the original center, but no translation yet
rotation_matrix = cv2.getRotationMatrix2D(center, angle, 1.0)

# Adjust the rotation matrix to take into account the translation
rotation_matrix[0, 2] += (new_w - w) // 2  # Shift in x direction
rotation_matrix[1, 2] += (new_h - h) // 2  # Shift in y direction

# Perform the rotation with the larger output size to avoid cropping
rotated_image = cv2.warpAffine(image, rotation_matrix, (new_w, new_h))

# Show the rotated image (or save it)
cv2.imshow('Rotated Image-fitted', rotated_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# We are copying the original image,
# as it is an in-place operation.
output = image.copy()

# Using the rectangle() function to create a rectangle.
rectangle = cv2.rectangle(output, (100, 100),
                        (400, 400), (255, 0, 0), 5)
cv2.imshow("Rectangle in image", output) 
cv2.waitKey(0)
cv2.destroyAllWindows()
```
